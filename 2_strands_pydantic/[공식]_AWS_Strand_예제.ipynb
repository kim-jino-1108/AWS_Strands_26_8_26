# Lab 1: Creating the Weather Agent




```python
from strands import Agent
from strands_tools import http_request
from strands.models.openai import OpenAIModel
```

```python
# 날씨 전용 시스템 프롬프트 정의
WEATHER_SYSTEM_PROMPT = """당신은 HTTP 요청 기능을 갖춘 날씨 어시스턴트입니다. 다음 작업을 수행할 수 있습니다:

1. 미국 국립기상청(National Weather Service) API로 HTTP 요청 보내기
2. 날씨 예보 데이터를 가공하고 표시하기
3. 미국 내 지역의 날씨 정보 제공하기

날씨 정보를 가져올 때:
1. 먼저 https://api.weather.gov/points/{latitude},{longitude} 또는 https://api.weather.gov/points/{zipcode} 를 사용해 좌표 또는 그리드 정보를 가져옵니다
2. 그런 다음 응답으로 반환된 forecast URL을 사용해 실제 예보를 가져옵니다

응답을 표시할 때:
- 날씨 데이터를 사람이 읽기 쉬운 형태로 정리합니다
- 기온, 강수, 기상 특보 등 중요한 정보를 강조합니다
- 오류는 적절히 처리합니다
- 전문 용어는 사용자가 이해하기 쉬운 표현으로 바꿉니다

항상 날씨 상황을 명확하게 설명하고, 예보에 대한 배경 설명을 함께 제공하세요.
"""

# 사용할 모델 식별자를 상수로 지정
MODEL = "gpt-5.4-mini"
# OpenAI 모델 객체 생성
# client_args: OpenAI 클라이언트에 전달할 인자, 환경 변수에서 API 키를 읽어 전달
# model_id: 위에서 지정한 모델 식별자
model = OpenAIModel(client_args={"api_key": os.environ["OPENAI_API_KEY"]}, model_id=MODEL)

# HTTP 요청 기능을 가진 에이전트 생성
# model: 위에서 생성한 모델 객체
# system_prompt: 에이전트의 역할과 동작 방식을 정의한 시스템 프롬프트
# tools: 에이전트가 사용할 도구 목록, http_request 도구를 명시적으로 활성화
weather_agent = Agent(
    model=model,
    system_prompt=WEATHER_SYSTEM_PROMPT,
    tools=[http_request],
)
```

```python
# 에이전트에게 질문을 전달하고 응답을 받음
# 좌표 조회, forecast URL 호출 등 API 세부 처리는 에이전트가 알아서 수행
# 반환값: 에이전트가 생성한 최종 응답 객체
response = weather_agent("시애틀 날씨는 어때?")
```

# Lab 2: 기본 에제

https://strandsagents.com/


```python
from strands import Agent, tool
from strands.hooks import BeforeToolCallEvent
from pathlib import Path
from strands.models.openai import OpenAIModel

# 사용할 모델 식별자를 상수로 지정
MODEL = "gpt-5.4-mini"
# OpenAI 모델 객체 생성
# client_args: OpenAI 클라이언트에 전달할 인자, 환경 변수에서 API 키를 읽어 전달
# model_id: 위에서 지정한 모델 식별자
model = OpenAIModel(client_args={"api_key": os.environ["OPENAI_API_KEY"]}, model_id=MODEL)


# @tool 데코레이터: 이 함수를 에이전트가 호출할 수 있는 도구로 등록
# 함수 시그니처와 docstring이 도구 이름/설명/입력 스키마로 변환됨
@tool
def save_report(title: str, content: str) -> str:
    """Save a research report to disk."""
    # 저장 경로 생성, reports 디렉터리 아래에 title.md 파일로 지정
    path = f"reports/{title}.md"
    # content 문자열을 해당 경로에 파일로 기록
    Path(path).write_text(content)
    # 도구 실행 결과로 모델에게 돌려줄 문자열 반환
    return f"Saved {path}"


# BeforeToolCallEvent 훅 콜백: 도구가 실제로 실행되기 직전에 호출됨
def require_sources(event: BeforeToolCallEvent):
    # 호출하려는 도구의 이름
    name = event.tool_use["name"]
    # 호출 인자(딕셔너리)를 문자열로 변환, 문자열 검사를 하기 위함
    inp = str(event.tool_use["input"])
    # 도구가 save_report이고 인자 안에 "[source]" 문자열이 없으면
    if name == "save_report" and "[source]" not in inp:
        # cancel_tool에 문자열을 대입하면 도구 실행이 취소되고
        # 그 문자열이 도구 결과 대신 모델에게 전달됨
        event.cancel_tool = "Add source citations."


# 에이전트 생성
# tools: 에이전트가 사용할 도구 목록
# hooks: 에이전트 실행 중 이벤트에 개입할 훅 목록
agent = Agent(
    model=model,
    tools=[save_report],
    hooks=[require_sources],
)
```

```python
agent("Research AI agent frameworks. You MUST call save_report with title='ai-agent-frameworks'. Every section of content must contain the literal marker [source](url).")
```
